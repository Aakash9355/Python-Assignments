Question 1 :  Write a Python program to check whether a number is even or odd using conditional statements.

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print(f"{num} is Even")
else:
    print(f"{num} is Odd")
```

Question 2 : Write a Python program to count how many times a number appears in a list.

Example List

[2,3,4,2,5,2]

```python
numbers = [2, 3, 4, 2, 5, 2]
target = 2

count_val = numbers.count(target)
print(f"{target} appears {count_val} times in the list.")
```

Question 7 : Find all records where:

Question 9 : Create a random array of 10 numbers and:
*  Find max and min
*  Sort the array





```python
import numpy as np

random_arr = np.random.randint(1, 100, size=10)
print("Random Array:", random_arr)

print("Maximum value:", np.max(random_arr))
print("Minimum value:", np.min(random_arr))

sorted_arr = np.sort(random_arr)
print("Sorted Array:", sorted_arr)
```

Question 10 : Replace all even numbers with 0:

arr = np.array([1,2,3,4,5,6])

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

arr[arr % 2 == 0] = 0
print("Modified Array:", arr)
```

```python
import numpy as np

arr_method2 = np.array([1, 2, 3, 4, 5, 6])

# Method 2: Using np.where
arr_method2 = np.where(arr_method2 % 2 == 0, 0, arr_method2)
print("Modified Array (Method 2):", arr_method2)
```
